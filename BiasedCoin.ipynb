```python
import pandas
import numpy
import matplotlib.pyplot as plt

def biased_coin(p,n):

    mu = 0
    for i in range(n):
        rand = numpy.random.default_rng()
        randi = rand.random()
        if randi < p:
            mu+=1
        else:
            mu+=0
    mu/=n
    return mu

```

```python
biased_coin(0.3,100)
means = numpy.array([])
for i in range(10000):
    result = biased_coin(0.3,100)
    means = numpy.append(means, result)

plt.hist(means)
```
