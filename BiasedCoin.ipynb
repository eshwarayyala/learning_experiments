```python
import numpy as np

def gDist(x,mu,si):
    return 1/(si*np.sqrt(2*np.pi))*np.exp((-(x-mu)**2)/(2*si**2))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def biased_coin(p,n):

    mu = 0
    for i in range(n):
        rand = np.random.default_rng()
        randi = rand.random()
        if randi < p:
            mu+=1
        else:
            mu+=0
    mu/=n
    return mu

```

```python
biased_coin(0.3,100)
```

```python
means = np.array([])
for i in range(10000):
    result = biased_coin(0.3,100)
    means = np.append(means, result)



si = np.sqrt(0.3 * (1-0.3) / 100)
mu = .3
x = np.linspace(mu-4*si,mu+4*si,1000)
y=gDist(x,mu,si)
plt.plot(x,y)

plt.hist(means,density=True)
plt.xlabel('Means')
plt.ylabel('Prob Density')
plt.show()
```
